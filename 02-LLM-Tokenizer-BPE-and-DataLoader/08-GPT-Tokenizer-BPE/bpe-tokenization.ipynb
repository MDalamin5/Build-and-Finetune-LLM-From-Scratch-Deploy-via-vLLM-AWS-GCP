## **GPT tokenizer `BPE`**

```python
import importlib
import tiktoken
print("tiktoken version: ", importlib.metadata.version("tiktoken"))
```

```python
tokenizer = tiktoken.get_encoding("gpt2")
```

## **Text to ides**

```python
text = (
    "Hello, do you like tea? <|endoftext|> In the sunlite terraces"
    "of someunknownPlace."
)

integers = tokenizer.encode(text, allowed_special={"<|endoftext|>"})
print(integers)
```

## **ids to text**

```python
tokenizer.decode(integers)
```

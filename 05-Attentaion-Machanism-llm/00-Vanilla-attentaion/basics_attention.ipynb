## Vanilla Attention

- Here we implement the **basic non-trainable model**.
- **Self-Attention:** In a sentence, it measures how each word is related to every other word.
- **Example:** `My name is Al Amin. He is an AI Engineer.`
- In this example, **"He"** is strongly related to **"Al Amin"**.
- Self-attention checks how **each individual word relates to all other words** in the sentence.
